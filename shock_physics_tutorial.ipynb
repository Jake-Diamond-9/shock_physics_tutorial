## Overview

This notebook accompanies the tutorial paper **"Shock Physics and the Plate Impact Experiment: A Tutorial"** and provides an interactive Python tool (`shock_physics_tutorial.py`) to compute the thermodynamic and kinematic states of condensed matter during **planar shock compression (Hugoniot)** and subsequent **isentropic release to zero pressure**.

## Key Assumptions 

* **Uniaxial Strain & Hydrodynamic Approximation:** Hydrostatic pressure dominates over deviatoric stresses ($p \equiv -t_{11} = -t_{22} = -t_{33}$); elastoplasticity, phase transitions, and melting are neglected.
* **Discontinuous & Adiabatic Shock:** The shock front is treated as a true discontinuity with no heat conduction or extrinsic heat sources.
* **Mie–Grüneisen EOS with Linear Hugoniot:** Assumes $U_S = C_0 + S\dot{x}$, constant specific heat at constant volume ($C^v$), and a volume-dependent Grüneisen parameter where $\gamma(v)/v = \gamma_R / v_R$ is constant.
* **Free-Surface Release:** Release state calculations assume expansion to zero pressure ($P = 0$), which strictly applies behind traction-free surfaces (and not necessarily at an unsymmetric impact interface).

## Calculation Modes & Inputs

The function `shock_physics_tutorial()` supports three calculation modes:

| Mode | Required Arguments | Description |
|---|---|---|
| **`"Impact"`** | `Velocity` ($\text{m/s}$), `ProjName`, `ProjThick` ($\text{m}$), `TargName`, `TargThick` ($\text{m}$) | Solves a flyer-plate impact experiment (e.g., a $1\text{ mm}$ $\text{Ta}$ flyer impacting a $5\text{ mm}$ $\text{Cu}$ target at $3000\text{ m/s}$). Returns two dictionaries: `proj, targ`. |
| **`"Pressure"`** | `Pressure` ($\text{Pa}$), `Duration` ($\text{s}$), `TargName`, `TargThick` ($\text{m}$) | Directly specifies the target shock pressure $p^+$ and initial shock pulse duration. Returns one dictionary: `targ`. |
| **`"Volume"`** | `VolumeRatio` ($v/v_R < 1$), `Duration` ($\text{s}$), `TargName`, `TargThick` ($\text{m}$) | Directly specifies the compressed volume ratio $J^+ = v^+/v_R$ and initial pulse duration. Returns one dictionary: `targ`. |

*(Optional arguments `ProjTemp` and `TargTemp` default to $298.0\text{ K}$; `Verbose=True` prints formatted results to the console).*
